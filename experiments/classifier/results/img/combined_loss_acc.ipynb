```python
import pandas as pd
import numpy as np
import importlib
import sys
import os
# Add the folder and subfolder to the system path
sys.path.append(os.path.abspath(os.path.join('../../../')))
from AIedes.models import classificator_model



year="2020"
# load all the models and results in the current ending pkl files
dir = '../res/'
files = [f for f in os.listdir(dir) if f.endswith('.pkl') and f.startswith('results_and_models_')]
print(f"Found {len(files)} files matching the criteria.")
# load the models and results for each input variable example of file results_and_models_T_TP_2020_[20, 20, 20]_temp_tp
res_all = []
for file in files:
    input_var = file.split('_')[-1]  # Extract the input variable from the filename
    print(f"Processing file: {file} with input variable: {input_var}")
    if input_var == "temp.pkl":
        res = pd.read_pickle(dir + file)
        res["input_var"] = "temp"
    elif input_var == "tp.pkl":
        res = pd.read_pickle(dir + file)
        res["input_var"] = "temp_tp"
    res_all.append(res)
# Concatenate all results into a single DataFrame
merged_results = pd.concat(res_all, ignore_index=True)
merged_results
```

```python
merged_results["input_var"].unique()
# rename the input var "temp" to "Temp." and "temp_tp" to "Temp. + Prec."
merged_results["input_var"] = merged_results["input_var"].replace({"temp": "Temp.", "temp_tp": "Temp. + Prec."})
```

```python

import seaborn as sns
import matplotlib.pyplot as plt

res_t = merged_results[merged_results['input_var'] == 'Temp.'].reset_index(drop=True)
res_t_tp = merged_results[merged_results['input_var'] == 'Temp. + Prec.'].reset_index(drop=True)

# --- 1) GLOBAL STYLE FOR BOTH PANELS ---
sns.set_theme(
    style="whitegrid",
    context="paper",
    font_scale=1.2,
    rc={
        "axes.labelsize": 12,
        "axes.titlesize": 14,
        "xtick.labelsize": 10,
        "ytick.labelsize": 10,
        "legend.fontsize": 10,
        "grid.linewidth": 0.5,
        "grid.color": "0.8"
    }
)

fig, axs = plt.subplots(1, 2, figsize=(15, 5))

# ================== LEFT PANEL: LOSS CURVES ==================
model_index = 13
res_df = res_t_tp.copy()
if model_index < len(res_df):
    md = res_df.iloc[model_index]
    epochs = range(1, len(md["train_losses"][5:]) + 1)

    axs[0].plot(epochs, md["train_losses"][5:], label='Training Loss',
                color='#2E86AB', linewidth=2, alpha=0.85)
    axs[0].plot(epochs, md["test_losses"][6:],   # shift for alignment
                label='Testing Loss', color='#F24236',
                linewidth=2, linestyle='--', alpha=0.85)

    axs[0].set_xscale("log")
    axs[0].set_xlabel('Epoch')
    axs[0].set_ylabel('Loss')
    #axs[0].set_title("Training & Testing Loss")
    axs[0].set_ylim(0, 1.2 * max(max(md["train_losses"][5:]), max(md["test_losses"][6:])))
    axs[0].legend(
        #title=f"Model Arch: {md['hidden_layers']} \nInput Var: temp+tp",
        frameon=True, loc="best"
    )
else:
    axs[0].text(.5, .5, "Model index out of range",
                ha='center', va='center')

# ================== RIGHT PANEL: PARAMS vs ACCURACY ==================
# prepare merged_results / deepness etc. as before…
res_t['input_var'] = 'Temp.'
merged = pd.concat([res_t, res_t_tp], ignore_index=True)
merged["number of hidden layers"] = merged["hidden_layers"].apply(len)

palette = {"Temp.": "#1f77b4", "Temp. + Prec.": "#ff7f0e"}
markers = {0: "o", 1: "s", 2: "D", 3: "X", 4: "P"}
# Nice display names for legend
input_labels = {
    "Temp.": "Temperature",
    "Temp. + Prec.": "Temperature + Precipitation"
}
# Build depth labels from the data present
depth_labels = {
    d: (f"{d} layer" if d == 1 else ("Logistic Regression" if d == 0 else f"{d} layers"))
    for d in sorted(merged["number of hidden layers"].unique())
}

# Display columns used only for plotting (controls legend labels)
merged["Input Variables"] = merged["input_var"].map(input_labels)
merged["Hidden layers"] = merged["number of hidden layers"].map(depth_labels)

# Orders to control legend order
hue_order = [input_labels["Temp. + Prec."], input_labels["Temp."]]
style_order = [depth_labels[d] for d in sorted(depth_labels)]

# Palettes/markers mapped to the new display names
palette_disp = {input_labels[k]: v for k, v in palette.items()}
markers_disp = {depth_labels[k]: v for k, v in markers.items() if k in depth_labels}

ax = axs[1]
sns.scatterplot(
    data=merged,
    x="num_params", y="MEAN_ACC_test",
    hue="Input Variables", hue_order=hue_order, palette=palette_disp,
    style="Hidden layers", style_order=style_order, markers=markers_disp,
    s=100, edgecolor="k", ax=ax
)
sns.lineplot(
    data=merged,
    x="num_params", y="MEAN_ACC_test",
    hue="Input Variables", hue_order=hue_order, palette=palette_disp,
    linewidth=1.5, dashes=True, legend=False, ax=ax
)
for (var, depth), grp in merged.groupby(["input_var", "number of hidden layers"]):
    ax.errorbar(
        grp["num_params"], grp["MEAN_ACC_test"],
        yerr=grp["STD_ACC_test"], fmt="none",
        capsize=4, ecolor=palette[var], alpha=0.8
    )

ax.set_xscale("log")
ax.set_xlabel("Number of Parameters (log scale)")
ax.set_ylabel("Test Accuracy")

# One combined legend with custom section titles (from column names)
# If you want a global legend title, set title="Your title" here.
#ax.legend(title="", frameon=True, loc='best')

from matplotlib.lines import Line2D

# Remove auto legend
leg = ax.get_legend()
if leg is not None:
    leg.remove()

# Filled markers for Inputs (hue)
hue_labels = hue_order
hue_handles = [
    Line2D([0], [0], marker='o', linestyle='',
           markersize=8,
           markerfacecolor=palette_disp[name],
           markeredgecolor='black',
           label=name)
    for name in hue_labels
]

# Hollow markers for Hidden layers (style)
style_labels = style_order
style_handles = [
    Line2D([0], [0], marker=markers_disp[name], linestyle='',
           markersize=8,
           markerfacecolor='none',
           markeredgecolor='black',
           label=name)
    for name in style_labels
]

x_leg = 0.6

# First legend (top)
leg_inputs = ax.legend(
    handles=hue_handles, labels=hue_labels,
    title="Input Variables", frameon=True,
    loc='upper left', bbox_to_anchor=(x_leg, 0.5),
    borderaxespad=0., ncol=1
)
ax.add_artist(leg_inputs)

# Second legend directly below the first (same x anchor)
fig.canvas.draw()  # ensure we have a renderer
renderer = fig.canvas.get_renderer()
bbox_disp = leg_inputs.get_window_extent(renderer=renderer)
bbox_axes = ax.transAxes.inverted().transform_bbox(bbox_disp)
gap = 0.02  # small vertical gap in Axes coords
y2 = bbox_axes.y0 - gap

ax.legend(
    handles=style_handles, labels=style_labels,
    title="Hidden layers", frameon=True,
    loc='upper left', bbox_to_anchor=(x_leg, y2),
    borderaxespad=0., ncol=1
)

# --- 3) APPLY IDENTICAL SPINE & GRID TWEAKS TO BOTH AXES ---
for a in axs:
    # remove top/right spines
    sns.despine(ax=a, trim=False)
    # thicken bottom/left
    a.spines['left'].set_linewidth(1.3)
    a.spines['bottom'].set_linewidth(1.3)
    # ensure grid
    a.grid(True, alpha=0.3, linestyle='-', linewidth=0.5)

# Add panel labels ("a", "b")
axs[0].text(
    0.05, 1.0, "a", transform=axs[0].transAxes,
    fontsize=16, fontweight='bold', va='top', ha='right'
)
axs[1].text(
    0.05, 1.0, "b", transform=axs[1].transAxes,
    fontsize=16, fontweight='bold', va='top', ha='right'
)

fig.tight_layout()
fig.savefig("./combined_loss_acc.pdf", dpi=300,
            bbox_inches='tight', facecolor='white')
plt.show()
```
